```python
import sqlite3
import pandas as pd    
from pathlib import Path        
```

```python
caminho_db = Path("./db/projeto_vendas.db")

conexao = sqlite3.connect(caminho_db)

query_vendas = """
SELECT DISTINCT * FROM vendas
"""

query_clientes = """
SELECT * FROM clientes
"""

query_produtos = """
SELECT * FROM produtos
"""

vendas = pd.read_sql_query(query_vendas, conexao)
clientes = pd.read_sql_query(query_clientes, conexao)
produtos = pd.read_sql_query(query_produtos, conexao)

conexao.close()

display(vendas)
display(clientes)
display(produtos)
```
